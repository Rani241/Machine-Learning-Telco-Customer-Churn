## 4. Analisis dan Perbandingan Karakteristik Klasifikasi

Berdasarkan implementasi di atas, berikut adalah perbedaan mendasar dari ketiga jenis klasifikasi:

### 1. Klasifikasi Biner (Binary Classification)
* **Karakteristik Target:** Ruang label hanya terdiri dari tepat dua kemungkinan nilai yang saling berlawanan (misal: Positif/Negatif, 0/1, atau Rock/Hip-Hop).
* **Fungsi Aktivasi & Output:** Pada pemodelan jaringan saraf atau regresi logistik, klasifikasi biner menghasilkan satu nilai probabilitas tunggal $p \in [0, 1]$ menggunakan fungsi aktivasi **Sigmoid**. Jika nilai $p \ge 0.5$, data diklasifikasikan ke kelas positif.
* **Format Data Target ($y$):** Berbentuk vektor 1 dimensi (vektor kolom tunggal) dengan jumlah kelas unik sebanyak 2.

### 2. Klasifikasi Multikelas (Multi-class Classification - Logika OR)
* **Karakteristik Target:** Ruang label memiliki lebih dari dua kategori ($K > 2$), namun berlaku aturan *mutual exclusivity* (setiap baris data hanya boleh menjadi anggota dari tepat satu kelas).
* **Logika OR:** Kondisi data dievaluasi dengan logika pemisahan mutlak: data adalah Kelas A **ATAU** Kelas B **ATAU** Kelas C (contoh: biji kacang adalah `Seker` OR `Barbunya` OR `Bombay`).
* **Fungsi Aktivasi & Output:** Menggunakan fungsi aktivasi **Softmax** di lapisan akhir untuk menghasilkan distribusi probabilitas di seluruh kelas, di mana total seluruh probabilitas sama dengan 1 ($\sum p_i = 1$). Kelas dengan probabilitas tertinggi ($\text{argmax}$) yang dipilih.
* **Format Data Target ($y$):** Berupa satu kolom kategori label tunggal (atau matriks *one-hot encoded* dengan aturan tepat satu angka 1 per baris).

### 3. Klasifikasi Multilabel (Multi-label Classification - Logika AND)
* **Karakteristik Target:** Setiap baris sampel data dapat memiliki beberapa label target yang bernilai benar/aktif secara bersamaan (*non-exclusive*).
* **Logika AND:** Setiap label target dievaluasi secara independen dan tidak saling meniadakan (contoh: film kartun Disney dapat bergenre `Musical` = 1 **DAN** `Adventure` = 1 secara simultan; parfum dapat memiliki aroma `Woody` = 1 **DAN** `Citrus` = 1).
* **Fungsi Aktivasi & Output:** Pemodelan menggunakan fungsi aktivasi **Sigmoid secara independen** untuk setiap label kelas target (bukan Softmax), sehingga total probabilitas seluruh kolom tidak harus bernilai 1.
* **Format Data Target ($y$):** Berbentuk matriks multi-kolom biner ($N \times L$), di mana $L$ adalah jumlah label yang diprediksi dan tiap kolom bernilai 0 atau 1.

## 4. Perbedaan Klasifikasi Multi-class (OR) vs Multi-label (AND)

| Aspek | Multi-class Classification (Logika OR) | Multi-label Classification (Logika AND) |
| :--- | :--- | :--- |
| **Definisi Target** | Memilih **tepat satu** kelas dari lebih dari dua kategori yang tersedia[cite: 5]. | Satu baris data dapat memiliki **beberapa label aktif sekaligus** secara bersamaan[cite: 5]. |
| **Sifat Relasi** | *Mutually exclusive* (saling meniadakan)[cite: 5]. | *Non-exclusive* (independen, tidak saling meniadakan)[cite: 5]. |
| **Logika Boolean** | **Logika OR** (Data adalah Kelas A **ATAU** Kelas B **ATAU** Kelas C)[cite: 5]. | **Logika AND** (Data memiliki Label 1 **DAN** Label 2 **DAN** Label 3)[cite: 5]. |
| **Format Target ($y$)** | Vektor 1 dimensi / satu kolom kategori (misalnya: `['Seker', 'Barbunya', ...']`)[cite: 5]. | Matriks biner multi-kolom ($N \times L$), di mana tiap kolom bernilai `0` atau `1`[cite: 5]. |
| **Fungsi Aktivasi** | **Softmax** (total seluruh probabilitas kelas sama dengan satu: $\sum p_i = 1$). | **Sigmoid independen** di setiap node output label. |
| **Contoh Kasus Data** | Varietas biji kacang (`Dry Bean`: Seker OR Barbunya OR Cali)[cite: 5]. | Genre film Disney (`Disney Movies`: Musical AND Adventure AND Comedy)[cite: 5]. |

---

### Contoh Penerapan Nyata:

* **Multi-class (Logika OR):** Pada dataset biji kacang (*Dry Bean*), satu butir kacang hanya dapat digolongkan ke salah satu varietas saja: ia adalah `Seker` ATAU `Barbunya`, tidak mungkin menjadi dua varietas sekaligus dalam waktu bersamaan[cite: 5].
* **Multi-label (Logika AND):** Pada dataset animasi film Disney, satu judul film dapat memiliki beberapa label genre yang aktif bersamaan: film *Frozen* adalah `Musical` DAN `Adventure` DAN `Fantasy`[cite: 5]. Hal serupa berlaku pada aroma parfum, di mana satu formula dapat memiliki aroma `Woody` DAN `Citrus` secara simultan[cite: 5].
